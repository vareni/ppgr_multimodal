```python
all_meals = pd.read_csv('../splits/all_meals.csv')
all_meals.shape
```

```python
all_meals["Timestamp"] = pd.to_datetime(all_meals["Timestamp"])

df = all_meals.sort_values(["sub", "Timestamp"]).copy()

df["Timestamp_before"] = df.groupby("sub")["Timestamp"].shift(1)
df["Timestamp_after"] = df.groupby("sub")["Timestamp"].shift(-1)

df["Minutes_before"] = (
    (df["Timestamp"] - df["Timestamp_before"])
    .dt.total_seconds()
    .div(60)
)

df["Minutes_after"] = (
    (df["Timestamp_after"] - df["Timestamp"])
    .dt.total_seconds()
    .div(60)
)
```

```python
df.head(40)
```

```python
data["Timestamp"] = pd.to_datetime(data["Timestamp"])

cols_to_add = [
    "Timestamp",
    "Timestamp_before",
    "Timestamp_after",
    "Minutes_before",
    "Minutes_after"
]

data = data.merge(
    df[cols_to_add],
    on="Timestamp",
    how="left"
)
```

```python
data = pd.read_csv('../splits/lunch_dinner/filter_before_60_after_120.csv')
data.columns
```

```python
CAP_MINUTES = 240

for col in ["Minutes_before", "Minutes_after"]:
    data[f"{col}_missing"] = data[col].isna().astype(int)

    data[f"{col}_capped"] = (
        data[col]
        .clip(upper=CAP_MINUTES)
        .fillna(CAP_MINUTES)
    )
```

```python
data.head()
```

```python
data.to_csv('../splits/lunch_dinner/filter_before_60_after_120_capped.csv', index=False)
```
